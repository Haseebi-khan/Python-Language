is vs ==
Explain the difference between is and == in Python. Provide a concrete example where two variables compare equal with == but are not the same object (is returns False). When would you intentionally use is?

````python
lista = [1,3,45]
listb = [1,3,45]

#  check value
print(lista == listb)
# check memory
print(lista is listb)
````
